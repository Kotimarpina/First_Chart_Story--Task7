### Objective
Create 3–4 simple charts where each chart answers one specific question, then connect them into one short data story.

```python
# Cell 1: Import Libraries
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
```

```python
# Cell 2: Upload the World Happiness Report dataset
uploaded = files.upload()

file_name = next(iter(uploaded))
print("Uploaded file:", file_name)
```

```python
# Cell 3: Load the dataset
df = pd.read_excel(file_name)

print("Dataset Shape:", df.shape)
df.head()
```

```python
# Cell 4: Understand the dataset
print("Column Names:")
print(df.columns.tolist())

print("\nDataset Information:")
df.info()
```

```python
# Cell 5: Check missing values and summary statistics
print("Missing Values:")
print(df.isnull().sum())

print("\nSummary Statistics:")
display(df.describe())
```

## Chart 1 – Which countries have the highest life evaluations?
**Chart type:** Horizontal bar chart

```python
# Cell 6: Top 10 Countries by Average Life Evaluation
top_10 = (
    df.groupby("Country name")["Life Ladder"]
      .mean()
      .sort_values(ascending=False)
      .head(10)
)

plt.figure(figsize=(10, 6))
top_10.sort_values().plot(kind="barh")
plt.title("Top 10 Countries by Average Life Evaluation")
plt.xlabel("Average Life Evaluation")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

**Takeaway:** The chart highlights the countries with the highest average reported life evaluations in the selected dataset.

## Chart 2 – Is economic well-being associated with happiness?
**Chart type:** Scatter plot

```python
# Cell 7: GDP vs Life Evaluation
gdp_data = df[["Log GDP per capita", "Life Ladder"]].dropna()

plt.figure(figsize=(9, 6))
plt.scatter(
    gdp_data["Log GDP per capita"],
    gdp_data["Life Ladder"],
    alpha=0.6
)
plt.title("Economic Well-being vs Life Evaluation")
plt.xlabel("Log GDP per Capita")
plt.ylabel("Life Evaluation")
plt.tight_layout()
plt.show()
```

**Takeaway:** The scatter plot helps examine whether countries with higher economic well-being also tend to report higher life evaluations.

## Chart 3 – Is social support associated with happiness?
**Chart type:** Scatter plot

```python
# Cell 8: Social Support vs Life Evaluation
social_data = df[["Social support", "Life Ladder"]].dropna()

plt.figure(figsize=(9, 6))
plt.scatter(
    social_data["Social support"],
    social_data["Life Ladder"],
    alpha=0.6
)
plt.title("Social Support vs Life Evaluation")
plt.xlabel("Social Support")
plt.ylabel("Life Evaluation")
plt.tight_layout()
plt.show()
```

**Takeaway:** The visualization helps examine whether stronger social support is associated with higher reported life evaluation.

## Chart 4 – Is freedom to make life choices associated with happiness?
**Chart type:** Scatter plot

```python
# Cell 9: Freedom vs Life Evaluation
freedom_data = df[["Freedom to make life choices", "Life Ladder"]].dropna()

plt.figure(figsize=(9, 6))
plt.scatter(
    freedom_data["Freedom to make life choices"],
    freedom_data["Life Ladder"],
    alpha=0.6
)
plt.title("Freedom to Make Life Choices vs Life Evaluation")
plt.xlabel("Freedom to Make Life Choices")
plt.ylabel("Life Evaluation")
plt.tight_layout()
plt.show()
```

**Takeaway:** The chart explores the relationship between perceived freedom to make life choices and reported life evaluation.

## Data Story
Happiness is influenced by more than a single factor. The analysis first identifies countries with higher reported life evaluations and then explores several factors associated with differences in well-being. Economic well-being is examined through GDP per capita, followed by social support and freedom to make life choices. Together, these visualizations provide a broader perspective on how different social and economic indicators relate to reported life evaluation across countries.
